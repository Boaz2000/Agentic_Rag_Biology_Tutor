```python
import os
from pathlib import Path
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings

print("📄 שלב 1: סורק את כל התיקיות וטוען קבצי PDF...")
loader = PyPDFDirectoryLoader("/content/drive/MyDrive/biology_data", recursive=True)
docs = loader.load()
print(f"✅ נטענו בהצלחה {len(docs)} עמודים.")

print("\n✂️ שלב 2: מחלק את הטקסט למקטעים (Chunks)...")
text_splitter = RecursiveCharacterTextSplitter(chunk_size=700, chunk_overlap=100)
chunks = text_splitter.split_documents(docs)
print(f"✅ הטקסט פוצל ל-{len(chunks)} מקטעים.")

print("\n🌐 שלב 3: טוען מודל שפה מקומי ומייצר את Chroma DB...")
# שימוש במודל רב-לשוני עוצמתי שמבין עברית מושלם ורץ ישירות על השרת של קולאב
embedding_model = HuggingFaceEmbeddings(model_name="intfloat/multilingual-e5-small")

db = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    persist_directory="./chroma_db"
)
print("✅ המחסן הוקטורי נוצר בהצלחה בתוך שניות!")

# 3. כיווץ התיקייה לקובץ ZIP
print("\n📦 מכווץ את התיקייה...")
!zip -r chroma_db.zip ./chroma_db
print("🏆 בינגו אחי! הקובץ chroma_db.zip מוכן ומחכה לך בלשונית התיקיות!")
```
